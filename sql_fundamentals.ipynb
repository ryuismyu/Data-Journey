## Exploring the dataset

```python
%%sql
# Shape and null check
SELECT COUNT(*) FROM customers;
SELECT COUNT(customer_id) FROM customers; # returns 26
```

```python
%%sql
SELECT MIN(customer_id), MAX(customer_id), AVG(customer_id)
FROM customers;
```

```python
%%sql
SELECT COUNT(*),
COUNT(DISTINCT city)
FROM customers;
```

```python
%%sql 
# 1. count orders by status
SELECT status,
COUNT(*)
FROM orders
GROUP BY status
ORDER BY COUNT(*) DESC;
```

```python
%%sql
# 2. Top customer
SELECT 
c.customer_id,
c.name,
COUNT(o.order_id) order_count
FROM customers c
INNER JOIN orders o 
    ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.name
ORDER BY order_count DESC
LIMIT 1;
```
