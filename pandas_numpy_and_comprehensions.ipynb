### Imoprtant concepts in python to review


## List vs Dict Indexing
- **List**: indexed by *position* (integer) → `passengers[0]` → the whole first item
- **Dict**: indexed by *key* (string/label) → `p['name']` → the value stored under that key
- `passengers` is a **list of dicts** — you can't use a string key directly on the list itself.
  You must first get *one dict* out (via position or a loop variable), *then* use a key on it.


## Comparison vs Assignment
- `=` **assigns** a value → `x = 5` means "set x to 5"
- `==` **compares** two values → `x == 5` returns `True` or `False`
- Conditions (`if`, filters) always need `==`, never `=`


## How `for` Loops Work
`for p in passengers:` — on each pass, `p` is *reassigned* to the next item in the list, in order:
- Pass 1: `p = passengers[0]`
- Pass 2: `p = passengers[1]`
- ...continues until the list runs out, then stops automatically.

Equivalent manual version:
​```python
i = 0
while i < len(passengers):
    p = passengers[i]
    i += 1
​```


## `return` vs `print`
- `return`: only valid **inside a function**; immediately **exits** the function/stops the loop
- `print`: displays output **without stopping** execution — use this inside a loop when you
  want to show something on *every* iteration


## List Comprehensions
Pattern: `[<what to keep> for <var> in <list> if <condition>]`

- Use a **comprehension** when you want a **new list back**
- Use a plain **`for` loop** when you want to **do something** (print, side effects) — no list needed
- Order matters: the `for` clause (defines the variable) must come before the `if` clause (which uses it)

Example:
​```python
those_survived = [p['name'] for p in passengers if p['survived'] == 1]
​```


## Why NumPy
- Python lists: flexible (mixed types allowed), but slow for numeric work — each element is a
  separate object; operations loop in Python, checking types one at a time
- NumPy arrays (`ndarray`): one contiguous block of memory, all elements the **same type**
- **Vectorization**: apply an operation to the *entire array at once* (fast, compiled code)
  instead of looping in Python
  - Loop version: `[age + 1 for age in ages]`
  - Vectorized version: `ages_array + 1`
- Trade-off: uniform type required — mixing `None` into a numeric array forces NumPy to either
  fall back to `dtype=object` (slow, defeats the purpose) or you use `np.nan` (the proper
  numeric missing-value marker) instead of `None`


## Basics of NumPy

````python
import numpy as np
````

## `None` vs `np.nan` in NumPy Arrays
- `np.array([29, 3, None])` → does **not** convert `None` to `0`
  - NumPy refuses to invent a fake numeric value (a missing age silently becoming `0` would be
    wrong data that looks valid — worse than an error, because nothing warns you)
  - Instead, since `int64` can't represent "missing," NumPy falls back to **`dtype=object`**
    (stores raw Python objects — `None` stays `None`, ints stay ints)
  - Downside: `dtype=object` loses vectorization — back to slow, Python-level looping

- **`np.nan`** ("Not a Number") is NumPy's actual missing-value sentinel for numbers
  - `np.array([29.0, 3.0, np.nan])` → stays in a fast numeric dtype: **`float64`**
  - Arithmetic **propagates** `nan` rather than skipping it: `.sum()` on an array containing
    `nan` returns `nan`, not a crash and not the sum of the valid values
  - To actually skip/ignore missing values in a calculation: use `np.nansum()`, `np.nanmean()`,
    etc. (the `nan`-aware versions)

**Takeaway:** `None` = generic Python "nothing" (breaks numeric arrays into slow object dtype).
`np.nan` = a real float value meaning "missing number" (keeps arrays fast, but math must be
`nan`-aware or results silently become `nan` too).


````python
np.array([29,3,None])
````

## None vs np.nan
- `np.array([29, 3, None])` → `dtype=object` (slow, no fake conversion to 0)
- `np.array([29.0, 3.0, np.nan])` → stays `float64` (fast)
- `nan` **propagates**: `.sum()`/`.mean()` on array with `nan` → `nan`
- `nan`-aware versions skip it: `np.nansum()`, `np.nanmean()`
- `nan == nan` is `False` — can't check for nan with `==`. Use `np.isnan(x)` instead


````python
# Fix
ages = np.array([29.0,3.0,np.nan])
ages.sum() # instead of adding 29.0 and 3 and treating sum as 32
# The sum just says np.float64(nan), numpy doesn't pretend to know the total
# and the whole result becomes nan.


# Average with np.nan in array

np.nanmean(ages) # This will ignore nan values first, and compute the mean of whatever's elft
````

## Boolean Masking
- `ages > 10` → array of `True`/`False`, one per element
- `ages[ages > 10]` → filters to just the elements where condition is `True`
- `nan` comparisons (`>`, `<`, `==`) always evaluate `False`
- Core idiom: `array[boolean_condition]` — same pattern Pandas filtering uses


````python
# Advantage of working with arrays
# instead fo running list comprehsnion filter, you can achieve simpler results 
# with numpy arrays

ages = np.array([29,45,3,61,np.nan])
ages>10 # gives array of true and false and can be filtered directly
# for true values
# nan comaprisons will always be False
ages[ages>10]
````

````python
# Returning only the non null values in a numpy array
ages[~np.isnan(ages)]
````

## Combining Conditions
- Use `&` (and) / `|` (or) — **not** Python's `and`/`or` (errors on arrays)
- Must parenthesize each comparison: `(ages > 10) & (ages < 50)`
- Function calls (e.g. `np.isnan(ages)`) don't need extra parens before `~`/`&`


````python
ages = np.array([29.0, 45.0, 3.0, 61.0, np.nan])   # Alice, Bob, Carla, Dan, Elena
survived = np.array([1, 0, 1, 0, 1])

ages[(~np.isnan(ages) & (survived==1))]
````

## Parallel Arrays (fragile!)
- Two separate arrays (`ages`, `survived`) are linked **only by matching index position**

- Pandas DataFrames fix this by keeping all columns bundled + aligned automatically

## Recap: GroupBy + Missing Data Handling

**GroupBy** — split → apply → combine
- `df.groupby('Sex')['Survived'].mean()` splits into groups, applies `.mean()` within each, combines results
- Multiple keys at once: `groupby(['Sex','Pclass'])` → labeled by a **MultiIndex** (tuple per row); pull one combo with `.loc['female', 1]`
- `.size()` = row count per group; `.mean()` = average per group — different questions, easy to mix up
- Why `.mean()` works on a 0/1 column: `mean = sum/count = (count of 1s)/total = rate` — same math as manually dividing survivors by total, just automatic

**Missing data — audit first, then decide per column:**
- `df.isna().sum() / len(df)` → % missing per column (percentage matters more than raw count)
- Decision depends on **% missing** and **dtype**:
  - `Embarked` (0.2% missing, categorical) → `dropna(subset=['Embarked'])`, cost is negligible
  - `Cabin` (77% missing, text) → drop the column; "filling" a mostly-missing text field has no
    sound basis (no mean/median for strings, no way to guess 77% of raw values)
  - `Age` (20% missing, numeric) → fill, but **not** with one global median — group by `Pclass`
    first (`groupby('Pclass')['Age'].transform('median')`) so the fill respects the real pattern
    (older/wealthier passengers cluster in 1st class) instead of flattening it

That's the same 0.6 you'd get doing "3 survivors ÷ 5 total" by hand — which is exactly the calculation you described back at the very start of this lesson. .mean() on a 0/1 column isn't a different concept, it's just your original divide-survivors-by-total logic computed automatically.

````python
# Using .dropna() to remove the data from the embarked column as it only represents 0.22% of the column
# and drop the column entirely

df_train = df_train.dropna(subset='Embarked')
````

````python
# Checking to see if 'Embarked' sstill has any nulls

df_train.isna().sum() # Embarked got 0 nulls
````

````python
# Now we need to address the age missing values, in which we will fill with the median age depending on pclass

#df_train.groupby('Pclass')['Age'].transform('median')
df_train['Age'] = df_train['Age'].fillna(df_train.groupby('Pclass')['Age'].transform('median'))
````

**Why this line needs both `transform` and `fillna`:**
- `.groupby('Pclass')['Age'].transform('median')` alone would overwrite **every** row's Age
  with its class median — even rows where Age is already known — because `transform` just
  broadcasts the group stat back to every row, it doesn't check for missing values at all
- `.fillna(...)` is what actually checks: if Age is `NaN`, use the value from the same row in
  the transform result; if Age is already known, leave it untouched
- Pandas' `median()` automatically skips `NaN`s when computing each group's median, so the
  fill values themselves aren't contaminated by the missing data being fixed

**`SettingWithCopyWarning`** (seen above): Pandas isn't sure if `df_train` at this point is a
fully independent DataFrame or a view tied to an earlier `dropna`/`drop` in the chain, so it
warns rather than risk a silent wrong result. It still worked here (confirmed below), but the
proper fix is adding `.copy()` right where `df_train` was last reassigned, to make it explicit
that it's independent — removes the ambiguity that triggers the warning.

### Using `.loc` and `.iloc`, which are for indexing

- `.loc` = label-based, here you give it a column name, like dictionary indexing
- `.iloc` = position-based, you give it an integer position counting from 0 , just like list indexing

````python
# Getting the first row with loc vs iloc

df_train.loc[1] # returns first row, passengerid (index) of 1

df_train.iloc[0] # also returns first row, idential to .loc[1]

df_train.loc[1,'Age'] # returns one specific cell by row and column name

df_train.iloc[0:5] # returns first 5 rows, same as df_train.loc[1:5]

df_train.loc[1:5] 
````

**Sharp edge: `.loc` slicing is inclusive, `.iloc` slicing is not.**
- `df_train.iloc[0:5]` → 5 rows (positions 0-4) — behaves exactly like regular Python list
  slicing (`list[0:5]` excludes the stop position)
- `df_train.loc[1:5]` → **also** 5 rows, but because it includes **both endpoints** (labels 1
  through 5) — label ranges work differently than position ranges
- Easy to get tripped up if you assume `.loc` follows the same exclusive-stop rule as `.iloc`
  or regular Python slicing — it doesn't.
